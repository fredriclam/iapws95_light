```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
import three_state_eos
model = three_state_eos.Model()
```

```python
rho_test_vec = np.concatenate((1/np.geomspace(1/800, 1e10, 200),
                                np.linspace(800, 1100, 1600)))
u_test_vec = np.linspace(-.6e6, 3e6, 801)
rho_mg, u_mg = np.meshgrid(rho_test_vec, u_test_vec)
rho_test = rho_mg.ravel()
u_test = u_mg.ravel()
```

```python
# Approximate thermodynamic variables over test range
out, region_labels_numeric = model.approx_rho_u(rho_test, u_test, extrapolate=False)
T, p, s, y_l, y_s = out.T
```

```python
# Contour of pressure
plt.contourf(rho_mg, u_mg, np.reshape(p, u_mg.shape))
plt.xlabel("Density (kg/m${}^3$)")
plt.ylabel("Specific energy (J/kg)")
plt.colorbar(label="Pressure (Pa)")
```

```python
# Contour of pressure
plt.contourf(rho_mg, u_mg, np.reshape(T, u_mg.shape))
plt.xlabel("Density (kg/m${}^3$)")
plt.ylabel("Specific energy (J/kg)")
plt.colorbar(label="Temperature (K)")
```
